PARA EXTRAER UNICAMENTE OHLC

```python
import MetaTrader5 as mt5
import pandas as pd

SYMBOL = "US500"
TIMEFRAME = mt5.TIMEFRAME_M15

def obtener_datos_ohlc(symbol, n_barras=99000):
    if not mt5.initialize():
        print("Error al inicializar MetaTrader5")
        return None
    
    print(f"Descargando datos OHLC de {symbol}...")
    rates = mt5.copy_rates_from_pos(symbol, TIMEFRAME, 0, n_barras)
    mt5.shutdown()
    
    if rates is None or len(rates) == 0:
        print("No se pudieron obtener datos.")
        return None
        
    df = pd.DataFrame(rates)
    
    # Convertir tiempo a formato legible
    df['time'] = pd.to_datetime(df['time'], unit='s')
    
    # Seleccionar solo las columnas necesarias (OHLC + time)
    # 'tick_volume', 'spread' y 'real_volume' se descartan
    df = df[['time', 'open', 'high', 'low', 'close']]
    
    return df

datos = obtener_datos_ohlc(SYMBOL)

if datos is not None:
    datos.to_csv(f"datos_ohlc_{SYMBOL}.csv", index=False)
    print(f"¡Completado! Archivo 'datos_ohlc_{SYMBOL}.csv' creado con {len(datos)} filas.")
else:
    print("Error: No se pudieron obtener los datos.")
    
```
